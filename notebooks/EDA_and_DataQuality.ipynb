# Exploratory Data Analysis & Data Quality

This notebook analyzes the Kathiyawadi market pricing dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set display options
pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
```

## 1. Data Loading
Loading the curated latest datasets.

```python
curated_dir = "../data/curated/latest"

if os.path.exists(curated_dir):
    df_restaurants = pd.read_csv(os.path.join(curated_dir, "Restaurant.csv"))
    df_products = pd.read_csv(os.path.join(curated_dir, "Product.csv"))
    df_facts = pd.read_csv(os.path.join(curated_dir, "Fact_Menu_Price.csv"))
    print(f"Loaded {len(df_restaurants)} restaurants, {len(df_products)} products, and {len(df_facts)} menu prices.")
else:
    print(f"Data not found at {curated_dir}. Make sure to run the pipeline first.")
```

## 2. Data Quality Checks
Checking for missing values, duplicates, and outliers.

```python
print("Missing Values in Restaurants:")
print(df_restaurants.isnull().sum())
print("\nMissing Values in Facts:")
print(df_facts.isnull().sum())
```

```python
df_merged = df_facts.merge(df_restaurants, on="Restaurant_ID", how="left")
df_merged = df_merged.merge(df_products, on="Product_ID", how="left")

# Check for price outliers globally vs per-dish
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_merged, x="Price")
plt.title("Global Price Distribution (Includes Outliers)")
plt.show()
```

## 3. Exploratory Data Analysis (EDA)
Analyzing price trends per dish.

```python
top_dishes = df_merged["Dish_Name"].value_counts().head(10).index

plt.figure(figsize=(12, 8))
sns.boxplot(data=df_merged[df_merged["Dish_Name"].isin(top_dishes)], y="Dish_Name", x="Price")
plt.title("Price Distribution for Top 10 Dishes")
plt.show()
```

Analyzing geographic trends by Area.

```python
plt.figure(figsize=(12, 6))
area_counts = df_restaurants["Area"].value_counts().head(10)
sns.barplot(x=area_counts.values, y=area_counts.index)
plt.title("Top 10 Areas with Most Restaurants")
plt.show()
```
